# каково соотношение побед к поражениям?

```python

win = wins_looses.loc[wins_looses["result"] == "Win", "count"].iloc[0]
cashout = wins_looses.loc[wins_looses["result"] == "Cashout", "count"].iloc[0]
lose = wins_looses.loc[wins_looses["result"] == "Lose", "count"].iloc[0]

print(f"Wins and cashouts: {win+cashout}")
print(f"Lose: {lose}")
print(f"Win/Lose: {win/lose}")
```

# длительность ставок

```python
sessions = bet

sessions["accept_time"] = pd.to_datetime(bet["accept_time"])
sessions["settlement_time"] = pd.to_datetime(bet["settlement_time"])

sessions['session_duration'] = bet['settlement_time']-bet['accept_time']
sessions = sessions [['bet_id','session_duration']].drop_duplicates()
sessions
```

# средняя и максимальная длительность ставки

```python
sessions_mean = sessions['session_duration'].mean()
sessions_max = sessions['session_duration'].max()
print(f"Средняя длительность ставки: {sessions_mean}")
print(f"Максимальная длительность ставки: {sessions_max}")
```

```python
# или можно датафреймом

sessions_mean2 = sessions
sessions_mean2 = sessions_mean2.agg(

    sessions_mean=("session_duration", "mean")

)
sessions_mean2
```

# игроки использущие только фриспины

```python
free_bet_players = (
    bet[['player_id', 'bet_id', 'is_free_bet']]
    .drop_duplicates()
    .groupby(['player_id', 'is_free_bet'])
    .size()
    .unstack(fill_value=0)
    .rename(columns={False: "non_free", True: "free"})
)

free_bet_players = free_bet_players[
    (free_bet_players["non_free"] == 0) &
    (free_bet_players["free"] > 10)
]

free_bet_players
```

# Customer Churn Analysis (процент халявщиков)

```python
freeloaders = free_bet_players.reset_index()
freeloaders = freeloaders['player_id'].nunique() / bet['player_id'].nunique() * 100

print (f'итого халявщиков: {freeloaders}')
```

# а они что-нибудь выигрывают? может жулики?

```python
# вариант 1 - merge()

profit_by_player = bet.groupby("player_id")["profit"].sum().reset_index()

freeloaders_profit = free_bet_players.merge(profit_by_player, on="player_id", how="left")
freeloaders_profit
```

```python
# вариант 2 - ищем данные в одной таблице по значениям из другой таблицы через isin()

profit_by_player = bet.groupby('player_id')['profit'].sum().reset_index()

freeloaders_profit2 = profit_by_player[profit_by_player['player_id'].isin(freeloaders_profit['player_id'])]
freeloaders_profit2
```
